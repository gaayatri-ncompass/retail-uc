Removing prefix from varchar columns


```python
location = pd.read_csv('../data/raw/source_erp/LOC_A101.csv')
location['CID'] = location['CID'].str.extract(r'(\d+)$').astype(int)

location.to_csv('../data/processed/erp/locations.csv', index=False)
```

```python
customers_erp = pd.read_csv('../data/raw/source_erp/CUST_AZ12.csv')
customers_erp['CID'] = customers_erp['CID'].str.extract(r'(\d+)$').astype(int)
customers_erp.to_csv('../data/processed/erp/customers_erp.csv', index=False)
```

```python
import re
customers_crm = pd.read_csv('../data/raw/source_crm/cust_info.csv')
customers_crm['cst_key'] = customers_crm['cst_key'].apply(
    lambda x: int(re.search(r'\d+$', x).group()) if pd.notnull(x) and re.search(r'\d+$', x) else x
)

customers_crm['cst_id'].dropna(inplace=True)

customers_crm.to_csv('../data/processed/crm/customers_crm.csv', index=False)
```

```python
sales_crm = pd.read_csv('../data/raw/source_crm/sales_details.csv')
sales_crm['sls_ord_num'] = sales_crm['sls_ord_num'].str.extract(r'(\d+)$').astype(int)
sales_crm.to_csv('../data/processed/crm/sales_crm.csv', index=False)
```

```python
import duckdb
import pandas as pd

# Load sample data
customers = pd.read_csv("../data/processed/crm/customers_crm.csv")
locations = pd.read_csv("../data/processed/erp/locations.csv")
birthdates = pd.read_csv("../data/processed/erp/customers_erp.csv")

con = duckdb.connect()
con.register("customers", customers)
con.register("customer_locations", locations)
con.register("customer_birthdates", birthdates)

query = """
 SELECT
            c.cst_id,
            c.first_name,
            c.last_name,
            CASE
                WHEN LOWER(c.gender) = 'f' THEN 'Female'
                WHEN LOWER(c.gender) = 'm' THEN 'Male'
                WHEN LOWER(b.gender) = 'females' THEN 'Female'
                WHEN LOWER(b.gender) = 'males' THEN 'Male'
                ELSE 'Unknown'
            END AS gender,
            c.marital_status,
            l.country,
            b.birthdate AS birth_date,
            c.created_date AS customer_created_date
        FROM customers c
        LEFT JOIN customer_locations l ON c.cst_id = l.cid
        LEFT JOIN customer_birthdates b ON c.cst_id = b.cid;
"""

result = con.execute(query).fetchdf()
print(result.head())
```
